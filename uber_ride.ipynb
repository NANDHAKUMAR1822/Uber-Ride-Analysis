```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

uber_df = pd.read_csv('uber_data.csv')
print ("FIRST 5 ROWS")
print(uber_df.head())

print("\n LAST 5 ROWS")
print(uber_df.tail())

print("\n COLUMNS")
print(uber_df.columns)

print("\n INFORMATION")
print(uber_df.info())

print("\n SHAPE")
print(uber_df.shape)

print("\n DESCRIBE")
print(uber_df.describe())

print("\n DATA TYPES")
print(uber_df.dtypes)

print("\n SIZE ")
print(uber_df.size)
### 1
print("null values")
print(uber_df.isnull().sum())

print("\n duplicate values")
print(uber_df.duplicated().sum())
uber_df.drop_duplicates(inplace=True)

print("\n unique values")
print(uber_df.nunique())
### 2
uber_df.columns
uber_df.columns.str.replace('*','',regex=False).str.replace(' ','').str.lower().str.strip()
### 3
uber_df['PURPOSE*']=uber_df['PURPOSE*'].fillna("unknown")
uber_df['END_DATE*']= uber_df['END_DATE*'].fillna(uber_df['END_DATE*'].mode()[0])
uber_df['CATEGORY*']=uber_df['CATEGORY*'].fillna(uber_df['CATEGORY*'].mode()[0])
uber_df['START*']=uber_df['START_DATE*'].fillna(uber_df['START*'].mode()[0])
uber_df['STOP*']=uber_df['STOP*'].fillna(uber_df['STOP*'].mode()[0])
print("null values after cleaning")
print(uber_df.isnull().sum())
### 4
print("\n statistical summary")
print(uber_df.describe())
### 5
uber_df["START_DATE*"]=pd.to_datetime(uber_df["START_DATE*"],errors = "coerce")
uber_df["END_DATE*"]=pd.to_datetime(uber_df["END_DATE*"],errors ="coerce")
print(uber_df["START_DATE*"])
print(uber_df["END_DATE*"])
### 6
uber_df["hour"]= uber_df["START_DATE*"].dt.hour
uber_df["day"]= uber_df["START_DATE*"].dt.day
uber_df["month"]= uber_df["START_DATE*"].dt.month
uber_df["year"]= uber_df["START_DATE*"].dt.year
uber_df["day_name"]= uber_df["START_DATE*"].dt.day_name()
print("\n updated dataset")
print(uber_df.head())
### 7
print("\n unique categories")
print(uber_df['CATEGORY*'].unique())

print("\n unique purposes")
print(uber_df['PURPOSE*'].unique())

print("\n unique starting locations")
print(uber_df['START*'].unique())
### 8
print("\n number of rides by category \n")
print(uber_df['CATEGORY*'].value_counts())

print("\n number of rides by purpose \n")
print(uber_df['PURPOSE*'].value_counts())

print("\n number of rides by starting location\n ")
print(uber_df['START*'].value_counts().head(10))

print("\n number of rides by ending location\n ")
print(uber_df['STOP*'].value_counts().head(10))
### 9
print("\n total miles\n ")
print(uber_df['MILES*'].sum())

print("\n average miles\n")
print(uber_df['MILES*'].mean)

print("\n maximum miles\n")
print(uber_df['MILES*'].max())

print("\n minimum miles\n")
print(uber_df['MILES*'].min())
### 10
print("\n average miles by category\n")
total_miles_by_category = uber_df.groupby('CATEGORY*')['MILES*'].mean()
print(total_miles_by_category)

print("\n total miles by category\n")
total_miles_by_category = uber_df.groupby('CATEGORY*')['MILES*'].sum()
print(total_miles_by_category)

print("\n average miles by purpose\n")
total_miles_by_purpose = uber_df.groupby('PURPOSE*')['MILES*'].mean()
print(total_miles_by_purpose)

print("\n total miles by purpose\n")
total_miles_by_purpose = uber_df.groupby('PURPOSE*')['MILES*'].sum()
print(total_miles_by_purpose)
###  11
month_count = uber_df['month'].value_counts().sort_index()
print("\n rides by month\n")
print(month_count)

hour_count = uber_df['hour'].value_counts().sort_index()
print("\n rides by hour\n")
print(hour_count)
### 12
plt.figure(figsize=(10, 6))
sns.countplot(data=uber_df, x="CATEGORY*")
plt.title("Number of Rides by Category")
plt.xlabel("Category")
plt.ylabel("Number of Rides ")
plt.show()
print("\n\n\n")

plt.figure(figsize=(10, 6))
sns.countplot(data=uber_df, y="PURPOSE*")
plt.title("Number of Rides by Purpose")
plt.xlabel("Purpose")
plt.ylabel("Number of Rides ")
plt.show()
### 13
plt.figure(figsize=(10, 6))
sns.countplot(data=uber_df,y="PURPOSE*",order=uber_df['PURPOSE*'].value_counts().index)
plt.title("Number of Rides by Purpose")
plt.ylabel("Purpose")
plt.xlabel("Number of Rides ")
plt.show()
### 14
plt.figure(figsize=(10, 6))
sns.countplot(data=uber_df,x="day_name",order=["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"], palette = 'muted')
plt.title("Number of Rides by Day \n")
plt.xlabel("Day ")
plt.ylabel("Number of Rides ")
plt.xticks(rotation=30)
plt.show()

plt.figure(figsize=(10, 10))
sns.violinplot(data=uber_df,x="PURPOSE*",y="MILES*")
plt.title("Distribution of Miles by Purpose")
plt.xlabel("Purpose")
plt.ylabel("Miles")
plt.xticks(rotation=90)
plt.show()
### 15
plt.figure(figsize=(10, 6))
sns.countplot(data=uber_df,x="hour",palette="tab20")
plt.title("Number of Rides by Hour \n")
plt.xlabel("Hour ")
plt.ylabel("Number of Rides ")
plt.show()
### 16
plt.figure(figsize=(10, 6))
sns.barplot(x=total_miles_by_category.index, y=total_miles_by_category.values)
plt.title("Total Miles by Category")
plt.xlabel("Category")
plt.ylabel("Total Miles")
plt.show()
### 17
category_counts= uber_df['CATEGORY*'].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(category_counts.values, labels=category_counts.index,autopct="%1.1f%%",startangle=90)
plt.title("Uber Rides by Category")
plt.show()
### 18
plt.figure(figsize=(10,6))
sns.scatterplot(data=uber_df,x="hour",y="MILES*")
plt.title("Hours V/S Miles")
plt.xlabel("Hour")
plt.ylabel("Number of Miles")
### 19
plt.figure(figsize=(20,12))
sns.boxplot(data=uber_df,x="CATEGORY*",y="MILES*")
plt.title(" Miles Distribution by Category")
plt.xlabel("Category")
plt.ylabel("Miles")
plt.show()
### 20
numeric_data = uber_df[["MILES*","hour","day","month"]]
plt.figure(figsize=(10, 6))
sns.heatmap(numeric_data.corr(), annot=True, cmap="tab20")###coolwarm  ###bwr  ###Greens  plasma hsv   twilight
plt.title("Correlation Heatmap")
plt.show()
### 21
print("FINAL RESULTS")

print("Total Number of Rides ")
print(len(uber_df))

print("\n Total Miles")
print(round(uber_df['MILES*'].sum(),2))

print("\n Average Miles")
print(round(uber_df['MILES*'].mean(),2))
### 22
print("Total number of rides\n",len(uber_df))
print("Total miles\n",round(uber_df['MILES*'].sum(),2))
print("Average miles\n",round(uber_df['MILES*'].mean(),2))
print("Maximum miles\n",round(uber_df['MILES*'].max(),2))
print("Minimum miles\n",round(uber_df['MILES*'].min(),2))
print("Total no of rides by category\n")
print(uber_df['CATEGORY*'].value_counts())

print("\n Total no of rides by purpose")
print(uber_df['PURPOSE*'].value_counts())

print("\n Most starting location\n")
print(uber_df['START*'].value_counts().head(10))

print("\n Most ending location\n")
print(uber_df['STOP*'].value_counts().head(10))

print("Rides by month\n")
print(uber_df["month"].value_counts())

print("Busiest day\n")
print(uber_df['day_name'].value_counts())

print("Busiest hour\n")
print(uber_df['hour'].value_counts())





```
